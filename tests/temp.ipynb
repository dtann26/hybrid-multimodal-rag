```python
from pymilvus import connections, Collection, utility

# Connect to your Milvus instance
connections.connect(alias="default", host="localhost", port="19530")

# List all collections
for name in utility.list_collections():
    coll = Collection(name)
    
    # Get the vector field name (usually "vector" or "embedding")
    vector_fields = [f.name for f in coll.schema.fields if f.dtype.name.startswith("FLOAT_VECTOR")]
    
    for field in vector_fields:
        # Load the index info for this field
        index_info = coll.indexes
        
        for idx in index_info:
            if idx.field_name == field:
                print(f"Collection: {name}")
                print(f"  Field: {field}")
                print(f"  Index Type: {idx.params['index_type']}")
                print(f"  Metric: {idx.params.get('metric_type', 'N/A')}")
                print(f"  Params: {idx.params.get('params', {})}")
                print("-" * 40)
```
